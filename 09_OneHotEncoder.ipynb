```python
import numpy as np
import pandas as pd
import seaborn as sns
d=sns.load_dataset('tips')
d.head(2)
```

```python
x=d.iloc[:,2:]
```

```python
x.head(2)
```

```python
x['sex'].value_counts()
```

##OneHotEncoding using Pandas

```python
pd.get_dummies(x,columns=['smoker','day'])
```

##. K-1 OneHotEncoding

```python
pd.get_dummies(x,columns=['smoker','day'],drop_first=True)
```

##OneHotEncoding using Sklearn

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(x,x.iloc[:,-1],test_size=0.2,random_state=2)
X_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32)
X_train_new = ohe.fit_transform(X_train[['smoker','day','sex']])
X_test_new = ohe.transform(X_test[['smoker','day','sex']])
```
