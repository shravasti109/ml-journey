```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

df = sns.load_dataset('planets')
df.head()
```

```python
df.shape
```

```python
df.dropna(inplace=True)
df.head()
```

```python
# Constructing a new feature 'mass_per_distance' from mass and distance
df['mass_per_distance'] = df['mass'] / df['distance']
df.head()
```

```python
# Categorize orbital period into short, medium, long
def period_type(val):
    if val <= 10:
        return 0   # short period
    elif val > 10 and val <= 100:
        return 1   # medium period
    else:
        return 2   # long period

df['period_type'] = df['orbital_period'].apply(period_type)
df.head()
```

```python
df['period_type'].value_counts()
```

```python
df.drop(columns=['orbital_period', 'mass', 'distance'], inplace=True)
df.head()
```

```python
# Look at method column
df['method'].value_counts()
```

```python
# EASY SPLIT 1 - Just create binary column
# Is it Transit method? Yes=1 No=0
df['is_transit'] = (df['method'] == 'Transit').astype(int)
df[['method', 'is_transit']].head(10)
```
