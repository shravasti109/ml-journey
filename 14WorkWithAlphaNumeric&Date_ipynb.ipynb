```python
import pandas as pd

# Sample dataframe
df = pd.DataFrame({
    'a': ['a512b', '12345', 'x93k2', '67890', 'm781z']
})

# Create empty columns
df['number_column'] = ""
df['alphabet_column'] = ""

for i in range(len(df['a'])):
    if df['a'][i].isdigit():
        df.loc[i, 'number_column'] = df['a'][i]
    else:
        df.loc[i, 'alphabet_column'] = df['a'][i]

print(df)
```

```python
import numpy as np
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'a': ['a512b', '12345', 'x93k2', '67890', 'm781z']
})

# Convert column to numpy array
arr = df['a'].values

# Create empty lists
number_col = []
alphabet_col = []

for value in arr:
    if value.isdigit():
        number_col.append(value)
        alphabet_col.append("")
    else:
        number_col.append("")
        alphabet_col.append(value)

# Add new columns to dataframe
df['number_column'] = number_col
df['alphabet_column'] = alphabet_col

print(df)
```

#working with dates

```python
import pandas as pd

data = {
    "date": [
        "2024-01-05",
        "2024-02-12",
        "2024-03-18",
        "2024-04-22",
        "2024-05-09",
        "2024-06-14",
        "2024-07-30",
        "2024-08-11",
        "2024-09-25",
        "2024-10-03"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['year'] = df['date'].dt.year
df['month_no.'] = df['date'].dt.month
df['month_name'] = df['date'].dt.month_name()
df['day'] = df['date'].dt.day
df['week'] = df['date'].dt.dayofweek
df['WeekName']=df['date'].dt.day_name()
#df['IsWeekend'] = np.where(df['WeekName'].isin(['Sunday', 'Saturday']), 1,0)
df['quarter'] = df['date'].dt.quarter
df['IsWeekend'] = np.where(df['WeekName'].isin(['Sunday', 'Saturday']), 1,0)
df['weekofyear'] = df['date'].dt.isocalendar().week
```
