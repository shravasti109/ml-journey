```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

# Load dataset
df = sns.load_dataset('iris')

# Features and target
X = df.drop(columns=['species'])
y = df['species']
```

```python
# Scale data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# ✅ BEFORE PCA - Train model on original 4 features
model = RandomForestClassifier()
scores_before = cross_val_score(model, X_scaled, y, cv=5, scoring='accuracy')

print("Accuracy BEFORE PCA:", scores_before)
print("Mean Accuracy BEFORE PCA:", round(np.mean(scores_before), 4))
```

```python
# Apply PCA - reduce to 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X_scaled.shape)
print("After PCA shape:", X_pca.shape)
print("Variance kept:", round(sum(pca.explained_variance_ratio_), 4))
```

```python
# ✅ AFTER PCA - Train model on 2 PCA features
scores_after = cross_val_score(model, X_pca, y, cv=5, scoring='accuracy')

print("Accuracy AFTER PCA:", scores_after)
print("Mean Accuracy AFTER PCA:", round(np.mean(scores_after), 4))
```
