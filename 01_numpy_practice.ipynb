```python
#2d array
import numpy as np
a=np.array([[1,2,3],[4,5,6]])
print(a)
print(a.ndim)
print(a.shape)
print(a.size)
print(a[0,1])
print(a.sum())
print(a.min())
print(a.max())
print(a.mean())
print(a.std())
print(a.var())
print(a.prod())
print(a.cumsum())
print(a.cumprod())
print(a.argmin())
print(a.argmax())
print(a.argsort())
print(a.reshape(3,2))
print(a.T)
print(a.flatten())
print(a.tolist())
```
